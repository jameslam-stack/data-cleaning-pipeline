# Data Cleaning Pipeline — Beginner Guide

This notebook teaches basic Python users how to use the `data-cleaning-pipeline` package.

**What you need:** basic Python (`print`, `for`, variables) and `pandas` (we use `pd.DataFrame`).


## Step 1 — Import libraries

Every Python script starts with imports. We import `pandas` as `pd` (the standard nickname) and our pipeline package.

```python
import pandas as pd
from data_cleaning_pipeline import DataCleaningPipeline, auto_clean_dataframe

print("Imports OK!")
```

## Step 2 — Make messy sample data

We create a DataFrame with common problems:
- duplicate rows
- missing values (`None`)
- wrong text case / extra spaces
- out-of-range ages

```python
df = pd.DataFrame({
    "id": [1, 2, 3, 4, 2],          # 2 is duplicate
    "name": ["Alice", "bob", "  CHARLIE  ", None, "Alice"],
    "age": [25, 30, 99, 20, 30],    # 99 is out of range
    "salary": [50000, 60000, 70000, 80000, 60000]
})

df
```

## Step 3 — Inspect the problem

Always look at the data before cleaning. Check:
- `df.info()` — column types
- `df.isnull().sum()` — missing values
- `df.duplicated().sum()` — duplicates

```python
print("Shape:", df.shape)
print("\nMissing:\n", df.isnull().sum())
print("\nDuplicates:\n", df.duplicated().sum())
```

## Step 4 — Use the pipeline (step by step)

We call methods in order. Each method returns `self`, so we can chain them with dots (`.`).

Order matters: **duplicates → missing → types → outliers → text**.

```python
pipeline = DataCleaningPipeline(df)

cleaned = (pipeline
    .remove_duplicates()           # drop duplicate rows
    .handle_missing_values()        # fill blanks with median / Unknown
    .fix_data_types()               # auto-convert dates / numbers
    .normalize_text()               # lowercase + trim whitespace
    .validate_ranges({"age": (0, 120)})  # keep only valid ages
    .get_cleaned_data()             # return the result
)

print("Cleaned shape:", cleaned.shape)
cleaned
```

## Step 5 — Read the cleaning report

The pipeline logs every step. Call `.generate_report()` to see a summary.

```python
print(pipeline.generate_report())
```

## Step 6 — Quick auto-clean (one line)

If you just want clean data fast, use `auto_clean_dataframe()`.

```python
quick = auto_clean_dataframe(df)
print("Quick cleaned:", quick.shape)
quick
```

## Step 7 — Save the result

Save to CSV (or JSON / Excel) with `.save_cleaned_data()`.

```python
pipeline.save_cleaned_data("cleaned_output.csv")
print("Saved to cleaned_output.csv")
```

## Recap — Key ideas for beginners

| Concept | Example in this notebook |
|---|---|
| Import | `import pandas as pd` |
| Create data | `pd.DataFrame({...})` |
| Check problems | `df.isnull().sum()` |
| Chain methods | `.remove_duplicates().handle_missing_values()` |
| Get result | `.get_cleaned_data()` |

**Tip:** If you get an error, read the message — it usually tells you which column or step failed.
