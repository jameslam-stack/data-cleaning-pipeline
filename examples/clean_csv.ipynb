# Clean a CSV File — File Workflow

This notebook shows the full workflow: load a messy CSV → clean → save.

**Prerequisite:** Run `teach_pipeline.ipynb` first (or read it).

```python
import pandas as pd
from data_cleaning_pipeline import DataCleaningPipeline, data_quality_checks

# Load the messy CSV
messy = pd.read_csv('messy_sample.csv')
print('Loaded messy CSV:', messy.shape)
messy.head()
```

```python
# Run quality checks BEFORE cleaning
print('=== BEFORE CLEANING ===')
data_quality_checks(
    messy,
    required_columns=['id','name','age','salary','email'],
    range_checks={'age':(0,120),'salary':(0,200000)},
    id_column='id',
    email_column='email',
)
```

```python
# Clean the data
pipeline = DataCleaningPipeline(messy)

cleaned = (pipeline
    .remove_duplicates()
    .handle_missing_values()
    .fix_data_types()
    .remove_outliers(columns=['age','salary'], method='iqr')
    .normalize_text()
    .validate_ranges({'age':(0,120),'salary':(0,200000)})
    .get_cleaned_data()
)

print('\n=== AFTER CLEANING ===')
cleaned
```

```python
# Run quality checks AFTER cleaning
print('=== AFTER CLEANING ===')
data_quality_checks(
    cleaned,
    required_columns=['id','name','age','salary','email'],
    range_checks={'age':(0,120),'salary':(0,200000)},
    id_column='id',
    email_column='email',
)
```

```python
# Save the cleaned data
pipeline.save_cleaned_data('cleaned_output.csv')
print('\nSaved cleaned_output.csv')
```

```python
# Verify by reloading
reloaded = pd.read_csv('cleaned_output.csv')
print('Reloaded:', reloaded.shape)
reloaded.head()
```

## Summary

| Step | Method | What it did |
|---|---|---|
| 1 | `pd.read_csv()` | Load file |
| 2 | `data_quality_checks()` | Diagnose problems |
| 3 | `.remove_duplicates()` | Drop duplicate rows |
| 4 | `.handle_missing_values()` | Fill blanks |
| 5 | `.fix_data_types()` | Fix types |
| 6 | `.remove_outliers()` | Remove bad values |
| 7 | `.normalize_text()` | Clean text |
| 8 | `.validate_ranges()` | Enforce limits |
| 9 | `.save_cleaned_data()` | Write result |

**Tip:** You can skip any step — the pipeline is flexible. Just don't skip `data_quality_checks()` to know what you're fixing.
